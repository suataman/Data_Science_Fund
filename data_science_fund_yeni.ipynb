```python
# 1. veri setini yükle
df = pd.read_csv("/content/e_ticaret_veri_seti.csv")
```

```python
# 4. eksik veri içeren satırları incele
eksik_olan_satirlar = df[df.isnull().any(axis = 1)]
eksik_olan_satirlar
```

```python
# 7. sayısal eksik verileri medyan ile doldurma
df_median = df.copy()
```

```python
say_sutun = df.select_dtypes(include = ["int64","float64"]).columns
say_sutun
```

```python
for sutun in say_sutun:
  df_median[sutun] = df_median[sutun].fillna(df_median[sutun].median())
df_median[say_sutun].isnull().sum()
```

```python
# yardımcı fonksiyon: metinleri normalize edelim
# büyük küçük harf dönüşümleri, boşlukların kaldırılması, Türkçe karakter kaynaklı farkların düzenlenmesi

def metin_normalize_et(series):

  return(
      series.astype(str)
      .str.strip()
      .str.lower()
      .str.replace("i̇", "i", regex=False)
      .str.replace("ı", "i", regex=False)
      .str.replace("ö", "o", regex=False)
      .str.replace("ü", "u", regex=False)
      .str.replace("ç", "c", regex=False)
      .str.replace("ş", "s", regex=False)
      .str.replace("ğ", "g", regex=False)
  )
```

```python
# 6. şehir sütunundaki tutarsız değerleri düzeltme
df["sehir"] = metin_normalize_et(df["sehir"])
df["sehir"].unique()
```

```python
df["kategori"] = metin_normalize_et(df["kategori"])
df["kategori"].unique()
```

```python
df["odeme_turu"] = metin_normalize_et(df["odeme_turu"])
df["odeme_turu"].unique()
```

```python
df["musteri_tipi"] = metin_normalize_et(df["musteri_tipi"])
df["musteri_tipi"].unique()
```

```python
kategori_normalize_map = {
    "ev ve yasam":"ev_yasam",
    "ev&yasam":"ev_yasam"
}
odeme_turu_normalize_map = {
    "kredi karti":"kredi_karti",
    "kart":"kredi_karti",
    "banka karti":"banka_karti",
    "kapida odeme":"kapida_odeme",
    "dijital cuzdan":"dijital_cüzdan",
    "havale/eft":"havale_eft",
    "eft":"havale_eft",
    "havale":"havale_eft",
}
```

```python
df["kategori"] = df["kategori"].replace(kategori_normalize_map)
df["kategori"].unique()
```

```python
df["odeme_turu"] = df["odeme_turu"].replace(odeme_turu_normalize_map)
df["odeme_turu"].unique()
```

## Aykırı Değerler (Outliers)

```python
import pandas as pd
```

```python
# 1. veri setini yüklemek
df = pd.read_csv("/content/e_ticaret_veri_seti_tekrarlayan_kayitlar_duzenlendi.csv")
```

```python
# 2. sayısal sütunların belirlenmesi
sayisal_sutun = df.select_dtypes(include=["int64","float64"]).columns
sayisal_sutun
```

```python
# 3. sayısal sütunların özet istatistiklerini inceleyelim
df[sayisal_sutun].describe()
```

```python
df.describe()
```

```python
"""

Outlier hesabı :
IQR = Q3(%75) - Q1(%25)
Üst sınır = Q3 + k.IQR
Alt sınır = Q1 - k.IQR

k : bizim belirlediğimiz bir sabit bu data setten sete değişir bu şimdilik 1 olsun

birim fiyat için :
IQR = 819 - 199 = 620
Üst sınır = 819 + (1*620) = 1439
Üst sınır = 199 - (1*620) = -421

bu iki değer arasında kalan değerlere aykkırı değerler yani Outlier denir
"""
```

```python
# 4. IQR ile aykırı değer sınırlarını hesaplama

aykiri_sinirlar = {}

for sutun in sayisal_sutun:

  q1 = df[sutun].quantile(0.25)
  q3 = df[sutun].quantile(0.75)

  iqr = q3 - q1
  alt_sinir = q1 - 1.5 * iqr
  ust_sinir = q3 + 1.5 * iqr

  aykiri_sinirlar[sutun] = {
      "q1": q1,
      "q3": q3,
      "iqr": iqr,
      "alt_sinir": alt_sinir,
      "ust_sinir": ust_sinir
  }
```

```python
aykiri_sinirlar
```

```python
# 5. aykırı değer içeren kayıtları tespit et
aykiri_maskesi = pd.Series(False, index = df.index)

for sutun in sayisal_sutun:
  alt_sinir = aykiri_sinirlar[sutun]["alt_sinir"]
  ust_sinir = aykiri_sinirlar[sutun]["ust_sinir"]

  sutun_maskesi = (df[sutun] < alt_sinir) | (df[sutun] > ust_sinir)
  aykiri_maskesi = aykiri_maskesi | sutun_maskesi
```

```python
aykiri_kayit = df[aykiri_maskesi]
```

```python
aykiri_kayit
```

```python
# 6. aykırı değer sayılarını sütun bazında incele

for sutun in sayisal_sutun:
  alt_sinir = aykiri_sinirlar[sutun]["alt_sinir"]
  ust_sinir = aykiri_sinirlar[sutun]["ust_sinir"]

  aykiri_sayi = ((df[sutun] < alt_sinir) | (df[sutun] > ust_sinir)).sum()
  print(f"{sutun}: {aykiri_sayi}")
```

```python
# 7. aykırı değer içeren kayıtları veri setinden temizleme
df_temiz = df[~aykiri_maskesi]
df_temiz.to_csv("e_ticaret_veri_seti_aykiri_degerler_duzenlendi.csv")
```

```python
# 3. kategoriye göre temel analiz
kategori_temel_analiz = df.groupby("kategori").agg(
    {
        "toplam_tutar" : ["sum","mean","count"] ,
        "birim_fiyat" : "mean",
        "musteri_puani":"mean",
        "teslimat_gunu": "mean"
    }
).round(1)

kategori_temel_analiz
```

## Pivot Tablo Analizi

```python
import pandas as pd
```

```python
# 1. veri setini yükle
df = pd.read_csv("/content/e_ticaret_veri_seti_aykiri_degerler_duzenlendi.csv")
df = df.loc[:, ~df.columns.str.contains('^Unnamed')]
df.head()
```

```python
# 2. kategori ve müşteri tipine göre pivot tablo analizi
# her kategori için müşteri tiplerine göre ortalama toplam tutar


pivot_kategori_musteri_tipi = pd.pivot_table(
    df,
    index = "kategori",
    columns = "musteri_tipi",
    values = "toplam_tutar",
    aggfunc = "mean"
)
pivot_kategori_musteri_tipi
```

```python
# 3. şehir ve ödeme türü göre pivot tablo analizi

sehir_odeme_turu_pivot = pd.pivot_table(
    df,
    index = "sehir",
    columns = "odeme_turu",
    values = "toplam_tutar",
    aggfunc = "mean"
).round(1)

sehir_odeme_turu_pivot
```

```python
# 4. kategori ve ödeme türüne göre pivot tablo analizi
kategori_odeme_turu_pivot = pd.pivot_table(
    df,
    index = "kategori",
    columns = "odeme_turu",
    values = "kargo_ucreti",
    aggfunc = "mean"
).round(1)

kategori_odeme_turu_pivot
```

## Zaman Serisi Analizi

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 1. veri setinin yüklenmesi
df.head()
```

```python
df.info()
```

```python
# 2. sipariş tarihini datetime tipine dönüştür
df["siparis_tarihi"] = pd.to_datetime(df["siparis_tarihi"])
df["siparis_tarihi"].dtype
```

```python
# 3. tarih sütunundan zaman özellikleri üretelim
df["siparis_yili"] = df["siparis_tarihi"].dt.year
df["siparis_ayi"] = df["siparis_tarihi"].dt.month
df["siparis_gunu"] = df["siparis_tarihi"].dt.day
df["haftanin_gunu"] = df["siparis_tarihi"].dt.day_name()
df["hafta_numarasi"] = df["siparis_tarihi"].dt.isocalendar().week.astype(int)
df.head()
```

```python
# 4. aylara göre sipariş yoğunluk

yillik_yogunluk = df.groupby("siparis_ayi").agg(
    siparis_sayisi = ("siparis_id","count"),
    benzersiz_müsteri_sayisi = ("musteri_id","nunique")
).reset_index().sort_values("siparis_ayi")

yillik_yogunluk
```

```python
plt.figure(figsize=(10,6))
plt.plot(yillik_yogunluk["siparis_ayi"],yillik_yogunluk["siparis_sayisi"],color = "red",marker ="o", label = "siparis_sayisi")
plt.plot(yillik_yogunluk["siparis_ayi"],yillik_yogunluk["benzersiz_müsteri_sayisi"],color = "green",marker ="o",label = "benzersiz_müsteri_sayisi")
plt.title("Ayalara göre Sipariş sayısı ve müşteri sayısı")
plt.xlabel("aylar")
plt.xticks(yillik_yogunluk["siparis_ayi"])
plt.tight_layout()
plt.legend()
plt.show()
```

```python
# 5. ayın ilk yarısı ve ikinci yarısında ki harcama farkını inceleme
df["ay_dönemi"] = df["siparis_gunu"].apply(
    lambda x: "Ayin_ilk_yarisi" if x<= 15 else "Ayin_ikinci_yarisi"
)

df.head()
```

```python
ay_dönemi_harcama = df.groupby("ay_dönemi").agg(
    döneme_gore_top_harcama = ("toplam_tutar","sum")
).reset_index().sort_values("ay_dönemi")

ay_dönemi_harcama
```

```python
plt.figure(figsize=(10,6))
sns.barplot(data= ay_dönemi_harcama, x= "ay_dönemi", y = "döneme_gore_top_harcama", palette="Greens_d")
plt.title("Ayın dönemine göre yapılan harcama")
plt.xlabel("ay_dönemi")
plt.ylabel("döneme_gore_top_harcama")
plt.tight_layout()
plt.show()
```

## Veri Hikayeleştirme ve İş Yorumu

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/e_ticaret_veri_seti_aykiri_degerler_duzenlendi.csv")
df = df.loc[:, ~df.columns.str.contains('^Unnamed')]
df.head()
```

```python
# 1. hangi kategori hem yüksek gelir hem de müşteri memnuniyetini yüksek sağlıyor

kategori_raporu = df.groupby("kategori").agg(
    toplam_gelir = ("toplam_tutar","sum"),
    musteri_memnuniyet = ("musteri_puani","mean")
).reset_index().round(2)
kategori_raporu
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# 1 satır, 2 sütunluk yan yana tuval kuruyoruz
# Yan yana duracakları için genişliği (ilk sayıyı) biraz daha büyük tutarız (örn: 14)
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 1. SOL TARAF: Toplam Ciro Grafiği
sns.barplot(
    ax=axes[0],
    data=kategori_raporu,
    x="kategori",
    y="toplam_gelir",
    palette="Blues_d",
)
axes[0].set_title("Kategorilere Göre Toplam Ciro", fontsize=12, fontweight="bold")
axes[0].tick_params(axis="x", rotation=15)

# 2. SAĞ TARAF: Müşteri Memnuniyeti Grafiği
sns.barplot(
    ax=axes[1],
    data=kategori_raporu,
    x="kategori",
    y="musteri_memnuniyet",
    palette="Purples_d",
)
axes[1].set_title(
    "Kategorilere Göre Müşteri Memnuniyeti", fontsize=12, fontweight="bold"
)
axes[1].tick_params(axis="x", rotation=15)

plt.tight_layout()
plt.show()
```
